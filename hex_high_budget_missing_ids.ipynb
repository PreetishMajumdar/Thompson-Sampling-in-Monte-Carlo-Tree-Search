```python
import pandas as pd

# Load your high-budget Hex results file
csv_file = "missing csf_hex_11x11_FINAL.csv"  # Update filename if needed

df = pd.read_csv(csv_file)
existing_ids = set(df["Game_ID"])
expected_ids = set(range(1, 101))  # Target dataset of 100 games

missing_ids = sorted(list(expected_ids - existing_ids))

print(f"Total Completed Games: {len(existing_ids)}/100")
print(f"Missing Games Count:  {len(missing_ids)}")
print(f"Missing Game IDs:     {missing_ids}")

# Save missing IDs to a text file for the filler script
with open("missing_ids.txt", "w") as f:
    for gid in missing_ids:
        f.write(f"{gid}\n")
print("\nSaved missing IDs to 'missing_ids.txt'")
```
